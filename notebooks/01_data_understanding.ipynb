```python
tables = {
    "accounts": df_accounts,
    "subscriptions": df_subscriptions,
    "feature_usage": df_feature_usage,
    "support_tickets": df_support,
    "churn_events": df_churn
}

for name, df in tables.items():
    print(f"--- {name} ---")
    print("Missing values:\n", df.isnull().sum())
    print("Duplicate rows:", df.duplicated().sum())
    print()
```

```python
for name, df in tables.items():
    print(name, "-> total missing values:", df.isnull().sum().sum(), "| duplicates:", df.duplicated().sum())
```

```python
for name, df in tables.items():
    print(f"--- {name} ---")
    print(df.dtypes)
    print()
```

```python
date_like_keywords = ["date", "_at"]

for name, df in tables.items():
    date_cols = [col for col in df.columns if any(k in col.lower() for k in date_like_keywords)]
    print(name, "->", date_cols)
```

```python
pd.to_datetime(df_accounts['signup_date']).min(), pd.to_datetime(df_accounts['signup_date']).max()
```

```python
date_columns = {
    "accounts": ("signup_date", df_accounts),
    "subscriptions_start": ("start_date", df_subscriptions),
    "subscriptions_end": ("end_date", df_subscriptions),
    "feature_usage": ("usage_date", df_feature_usage),
    "support_submitted": ("submitted_at", df_support),
    "support_closed": ("closed_at", df_support),
    "churn_events": ("churn_date", df_churn)
}

for name, (col, df) in date_columns.items():
    converted = pd.to_datetime(df[col])
    print(name, "->", converted.min(), "to", converted.max())
```
